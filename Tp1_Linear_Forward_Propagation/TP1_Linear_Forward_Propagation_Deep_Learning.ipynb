## **Linear_Forward_Propagation  -Deep_Learning**

**Code 1: Linear Process with 1 Neuron (3 Inputs $\rightarrow$ 1 Neuron)**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Input Data: 3 Clients x 3 Features [Income, Risk, Duration]
X = np.array([
    [0.9, 0.1, 0.9],  # Client 1
    [0.2, 0.5, 0.1],  # Client 2
    [0.5, 0.5, 0.5]   # Client 3
])                                    # Shape: (3, 3)
```

```python
# 2. Weights and Bias for 1 Neuron
W = np.array([[ 0.8],                 # Income weight
              [ 0.5],                 # Risk weight
              [-0.9]])                # Duration weight -> Shape: (3, 1)
b = 0.1

# 3. Linear Calculation: Z = X . W + b
Z = np.dot(X, W) + b

print(f"Input X shape:  {X.shape}")
print(f"Weight W shape: {W.shape}")
print(f"Output Z shape: {Z.shape}")
print(f"Linear Scores (Z):\n{Z}\n")
```

```python
# 4. ANN Figure (3 Inputs -> 1 Neuron)
fig, ax = plt.subplots(figsize=(9, 5))
ax.axis('off')

input_pos  = [(0, 2.0), (0, 1.0), (0, 0.0)]
neuron_pos = [(3.0, 1.0)]

input_labels  = ['X1\n(Income)', 'X2\n(Risk)', 'X3\n(Duration)']
neuron_labels = [f'N1\n(b={b})']

# Draw connections and weights
for i, (x1, y1) in enumerate(input_pos):
    x2, y2 = neuron_pos[0]
    ax.plot([x1, x2], [y1, y2], color='#4A6572', lw=1.8, zorder=1)
    mx, my = (x1 + x2) / 2, (y1 + y2) / 2
    ax.text(mx, my + 0.08, f'w{i+1} = {W[i, 0]}', fontsize=10, fontweight='bold',
            color='#1A237E', ha='center',
            bbox=dict(boxstyle='round,pad=0.2', fc='white', ec='none', alpha=0.85))

# Draw output arrow
ax.annotate('', xy=(4.4, 1.0), xytext=(3.32, 1.0),
            arrowprops=dict(arrowstyle='->', lw=2, color='#2E7D32'))
ax.text(4.5, 1.0, 'Z (3 x 1)', fontsize=11, fontweight='bold', color='#2E7D32', va='center')

# Draw nodes
for (x, y), label in zip(input_pos, input_labels):
    ax.add_patch(plt.Circle((x, y), 0.30, facecolor='#E3F2FD', edgecolor='#1565C0', lw=2, zorder=2))
    ax.text(x, y, label, ha='center', va='center', fontsize=9, fontweight='bold', zorder=3)

for (x, y), label in zip(neuron_pos, neuron_labels):
    ax.add_patch(plt.Circle((x, y), 0.32, facecolor='#FFF3E0', edgecolor='#E65100', lw=2, zorder=2))
    ax.text(x, y, label, ha='center', va='center', fontsize=10, fontweight='bold', zorder=3)

ax.text(0.0, 2.55, 'Input Layer\nX: (3 x 3)', ha='center', fontsize=11, fontweight='bold', color='#1565C0')
ax.text(3.0, 2.55, '1-Neuron Layer\nW: (3 x 1)', ha='center', fontsize=11, fontweight='bold', color='#E65100')

plt.title('Linear ANN Architecture: 1 Neuron (Z = X · W + b)', fontsize=13, fontweight='bold', pad=10)
ax.set_xlim(-0.7, 5.4)
ax.set_ylim(-0.5, 2.9)
plt.tight_layout()
plt.show()
```

**Code 2: Linear Process with 2 Neurons (3 Inputs $\rightarrow$ 2 Neurons in Parallel)**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Input Data: 3 Clients x 3 Features [Income, Risk, Duration]
X = np.array([
    [0.9, 0.1, 0.9],  # Client 1
    [0.2, 0.5, 0.1],  # Client 2
    [0.5, 0.5, 0.5]   # Client 3
])                                    # Shape: (3, 3)
```

```python
# 2. Weights and Bias for 2 Parallel Neurons
# Column 1 = Neuron 1 (Risk View), Column 2 = Neuron 2 (Growth View)
W = np.array([
    [ 0.8, 0.9],                      # Income weights
    [ 0.5, 0.2],                      # Risk weights
    [-0.9, 0.9]                       # Duration weights
])                                    # Shape: (3, 2)
b = np.array([[0.1, 0.1]])            # Shape: (1, 2)

# 3. Linear Calculation: Z = X . W + b
Z = np.dot(X, W) + b

print(f"Input X shape:  {X.shape}")
print(f"Weight W shape: {W.shape}")
print(f"Output Z shape: {Z.shape}")
print(f"Linear Scores (Z):\n{Z}\n")
```

```python
# 4. ANN Figure (3 Inputs -> 2 Parallel Neurons)
fig, ax = plt.subplots(figsize=(10, 5.5))
ax.axis('off')

input_pos  = [(0, 2.0), (0, 1.0), (0, 0.0)]
neuron_pos = [(3.0, 1.5), (3.0, 0.5)]

input_labels  = ['X1\n(Income)', 'X2\n(Risk)', 'X3\n(Duration)']
neuron_labels = [f'N1\n(b={b[0,0]})', f'N2\n(b={b[0,1]})']

# Draw connections and weights
for i, (x1, y1) in enumerate(input_pos):
    for j, (x2, y2) in enumerate(neuron_pos):
        ax.plot([x1, x2], [y1, y2], color='#4A6572', lw=1.6, zorder=1)
        t = 0.32 if j == 0 else 0.68
        mx, my = x1 + t * (x2 - x1), y1 + t * (y2 - y1)
        ax.text(mx, my + 0.06, f'{W[i, j]}', fontsize=9, fontweight='bold',
                color='#1A237E' if j == 0 else '#B71C1C', ha='center',
                bbox=dict(boxstyle='round,pad=0.15', fc='white', ec='none', alpha=0.85))

# Draw output arrows
for j, (x, y) in enumerate(neuron_pos):
    ax.annotate('', xy=(4.4, y), xytext=(3.32, y),
                arrowprops=dict(arrowstyle='->', lw=2, color='#2E7D32'))
    ax.text(4.5, y, f'Z_{j+1} (Col {j+1})', fontsize=10, fontweight='bold', color='#2E7D32', va='center')

# Draw nodes
for (x, y), label in zip(input_pos, input_labels):
    ax.add_patch(plt.Circle((x, y), 0.30, facecolor='#E3F2FD', edgecolor='#1565C0', lw=2, zorder=2))
    ax.text(x, y, label, ha='center', va='center', fontsize=9, fontweight='bold', zorder=3)

for (x, y), label in zip(neuron_pos, neuron_labels):
    ax.add_patch(plt.Circle((x, y), 0.32, facecolor='#FFF3E0', edgecolor='#E65100', lw=2, zorder=2))
    ax.text(x, y, label, ha='center', va='center', fontsize=10, fontweight='bold', zorder=3)

ax.text(0.0, 2.55, 'Input Layer\nX: (3 x 3)', ha='center', fontsize=11, fontweight='bold', color='#1565C0')
ax.text(3.0, 2.55, '2-Neuron Layer\nW: (3 x 2)', ha='center', fontsize=11, fontweight='bold', color='#E65100')

plt.title('Linear ANN Architecture: 2 Parallel Neurons (Z = X · W + b)', fontsize=13, fontweight='bold', pad=10)
ax.set_xlim(-0.7, 5.6)
ax.set_ylim(-0.5, 2.9)
plt.tight_layout()
plt.show()
```
